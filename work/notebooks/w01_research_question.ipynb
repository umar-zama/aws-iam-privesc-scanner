## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## 2. The Question

**Search question:** Which indexed pages are underperforming their expected CTR
for their position tier and content type — and are therefore worth an SEO
reviewer's time to investigate (title/snippet rewrite, internal linking, etc.)?

**Unit of analysis:** one page (one row = one content item, at its current
position tier and content type).

**Output:** a ranked list of pages, ordered by "CTR gap" — how far below the
expected CTR for their tier/type peers they are, weighted by how much traffic
is at stake (impressions).

**The decision this improves:** which pages a content/SEO reviewer should look
at first when triaging underperforming content, instead of reviewing pages in
an arbitrary order or by traffic volume alone.

**The action someone takes:** a reviewer opens the top-ranked pages and checks
title tags, meta descriptions, and snippet presentation for anything that could
be suppressing clicks despite decent visibility.

**Cost of a wrong recommendation:** if the score flags a page that's actually
fine, the team spends review time on a false positive — low cost, just wasted
hours. If the score misses a genuinely underperforming high-impression page,
that's a bigger cost: real traffic left on the table for longer, because no one
looked at it.

**Why data/ML helps at all:** with tens of thousands of pages, no reviewer can
manually compare each page's CTR against a fair peer group (same position tier,
same content type). A scoring model can do that comparison at scale and rank
pages by opportunity size, something a flat "sort by low CTR" rule can't do
fairly because it ignores that CTR naturally varies by position.
